# 📊 Smart Statistics & AI Search Tool

This tool allows you to search for statistics topics across trusted academic sources, YouTube channels, and AI agents (**Gemini & ChatGPT**).

---

```python
import webbrowser
import urllib.parse

def search_trusted_statistics(topic, choice):
    # Define queries using safe formatting
    yt_query = "{} (StatQuest OR ZedStatistics OR Khan Academy OR Professor Leonard)".format(topic)
    google_query = '"{}" site:openintro.org OR site:psu.edu OR site:khanacademy.org'.format(topic)
    
    # Define URLs
    yt_url = "https://www.youtube.com/results?search_query=" + urllib.parse.quote(yt_query)
    google_url = "https://www.google.com/search?q=" + urllib.parse.quote(google_query)
    
    urls = {
        "youtube": yt_url,
        "google": google_url,
        "gemini": "https://gemini.google.com/app",
        "chatgpt": "https://chatgpt.com/"
    }

    if choice == "1":
        print("Opening everything for {}...".format(topic))
        for url in urls.values():
            webbrowser.open(url)
    elif choice == "2":
        print("Opening YouTube...")
        webbrowser.open(urls["youtube"])
    elif choice == "3":
        print("Opening Google...")
        webbrowser.open(urls["google"])
    elif choice == "4":
        print("Opening AI Agents...")
        webbrowser.open(urls["gemini"])
        webbrowser.open(urls["chatgpt"])
    else:
        print("Defaulting to Google...")
        webbrowser.open(urls["google"])
```

```python
topic = input("Enter the statistics topic: ")

if topic.strip():
    print("\nWhere would you like to search?")
    print("1: ALL (YouTube, Google, Gemini, ChatGPT)")
    print("2: YouTube Only")
    print("3: Google Only")
    print("4: AI Agents (Gemini & ChatGPT)")
    
    choice = input("Enter number (1-4): ")
    search_trusted_statistics(topic, choice)
else:
    print("No topic provided.")
```
